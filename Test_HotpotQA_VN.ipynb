```python
def find_column(columns, candidates):
    lower_map = {c.lower(): c for c in columns}

    for candidate in candidates:
        if candidate.lower() in lower_map:
            return lower_map[candidate.lower()]

    return None


vn_id_col = find_column(
    vn_queries.column_names,
    ["id", "query-id", "query_id", "_id"]
)

vn_text_col = find_column(
    vn_queries.column_names,
    ["text", "query", "question"]
)

print("ID column:", vn_id_col)
print("Text column:", vn_text_col)
```

```python
def normalize_text(text):

    if text is None:
        return ""

    text = unicodedata.normalize(
        "NFKC",
        str(text)
    )

    text = text.lower()

    # chuẩn hóa khoảng trắng
    text = re.sub(r"\s+", " ", text)

    # bỏ khoảng trắng trước punctuation
    text = re.sub(
        r"\s+([.,!?;:])",
        r"\1",
        text
    )

    return text.strip()
```

```python
total_qrel_checked = 0

title_found_in_original_context = 0
exact_text_match = 0
text_containment_match = 0

content_mismatches = []


for qid, corpus_ids in qrels_by_query.items():

    original = original_by_id.get(qid)

    if original is None:
        continue

    # context của chính query đó
    context_map = {}

    for title, sentences in zip(
        original["context"]["title"],
        original["context"]["sentences"]
    ):

        context_text = " ".join(sentences)

        context_map[title] = context_text

    # kiểm tra từng qrel document
    for cid in corpus_ids:

        total_qrel_checked += 1

        doc = found_docs.get(cid)

        if doc is None:
            content_mismatches.append({
                "query_id": qid,
                "corpus_id": cid,
                "reason": "corpus_id_not_found"
            })
            continue

        og_title = doc["og_title"]
        og_text = doc["og_text"]

        # title có trong original query context không?
        if og_title not in context_map:

            content_mismatches.append({
                "query_id": qid,
                "corpus_id": cid,
                "reason": "title_not_in_original_context",
                "title": og_title
            })

            continue

        title_found_in_original_context += 1

        original_text = context_map[og_title]

        a = normalize_text(og_text)
        b = normalize_text(original_text)

        # 1. exact
        if a == b:
            exact_text_match += 1
            continue

        # 2. một bản có thể dài hơn một chút
        if a in b or b in a:
            text_containment_match += 1
            continue

        content_mismatches.append({
            "query_id": qid,
            "corpus_id": cid,
            "reason": "text_difference",
            "title": og_title,
            "green_text": og_text,
            "original_text": original_text
        })


print("=" * 70)
print("FULL CONTENT MAPPING CHECK")
print("=" * 70)

print("Total qrel rows checked :", total_qrel_checked)

print(
    "Title found in original:",
    title_found_in_original_context
)

print(
    "Exact text match        :",
    exact_text_match
)

print(
    "Containment match       :",
    text_containment_match
)

print(
    "Other mismatch          :",
    len(content_mismatches)
)
```

```python
def normalize_text_strict(text):
    if text is None:
        return ""

    text = unicodedata.normalize("NFKC", str(text))

    # Một số loại whitespace đặc biệt
    text = text.replace("\xa0", " ")

    # lowercase
    text = text.lower()

    # gom whitespace
    text = re.sub(r"\s+", " ", text)

    # bỏ space ngay sau dấu quote
    text = re.sub(r'"\s+', '"', text)

    # bỏ space ngay trước dấu quote
    text = re.sub(r'\s+"', '"', text)

    # bỏ space trước punctuation
    text = re.sub(r"\s+([.,!?;:])", r"\1", text)

    return text.strip()


strict_exact = 0
strict_mismatch = []

for x in content_mismatches:

    if x["reason"] != "text_difference":
        strict_mismatch.append(x)
        continue

    green = normalize_text_strict(x["green_text"])
    original = normalize_text_strict(x["original_text"])

    if green == original:
        strict_exact += 1
    else:
        strict_mismatch.append(x)


print("=" * 70)
print("STRICT NORMALIZATION CHECK")
print("=" * 70)

print("Original text differences :", len(content_mismatches))
print("Resolved by normalization :", strict_exact)
print("Still genuinely different :", len(strict_mismatch))
```

```python
from difflib import SequenceMatcher

print("Remaining mismatches:", len(strict_mismatch))

for i, x in enumerate(strict_mismatch[:29], start=1):

    green = normalize_text_strict(x["green_text"])
    original = normalize_text_strict(x["original_text"])

    similarity = SequenceMatcher(
        None,
        green,
        original
    ).ratio()

    print("\n" + "#" * 100)
    print("CASE", i)
    print("Query ID :", x["query_id"])
    print("Corpus ID:", x["corpus_id"])
    print("Title    :", x.get("title"))
    print(f"Similarity: {similarity:.4f}")

    print("\nGREENNODE:")
    print(x["green_text"])

    print("\nORIGINAL:")
    print(x["original_text"])
```

## Cell A1.

```python
import random
import re
import json
import unicodedata
from collections import Counter, defaultdict
from datasets import load_dataset

VN_REPO = "GreenNode/hotpotqa-vn"
EN_REPO = "hotpotqa/hotpot_qa"

SEED = 42
N_QUERIES = 1000

# Vietnamese
vn_queries = load_dataset(
    VN_REPO, "queries", split="test"
)

qrels = load_dataset(
    VN_REPO, "qrels", split="test"
)

# Original
hotpot_train = load_dataset(
    EN_REPO, "distractor", split="train"
)

hotpot_val = load_dataset(
    EN_REPO, "distractor", split="validation"
)

original_by_id = {}

for row in hotpot_train:
    original_by_id[str(row["id"])] = row

for row in hotpot_val:
    original_by_id[str(row["id"])] = row

vn_by_id = {
    str(row["id"]): row
    for row in vn_queries
}

# random cố định
all_ids = [str(x) for x in vn_queries["id"]]

rng = random.Random(SEED)
selected_ids = rng.sample(all_ids, N_QUERIES)

selected_id_set = set(selected_ids)

print("Selected queries:", len(selected_ids))
print("Seed:", SEED)
```

## Cell A2. lấy toàn bộ context của 1000 query

```python
def canonical(text):
    """
    Normalize mạnh để match GreenNode og_text
    với HotpotQA original dù khác whitespace.
    """
    if text is None:
        return ""

    text = unicodedata.normalize("NFKC", str(text))
    text = text.casefold()

    # loại toàn bộ whitespace
    text = re.sub(r"\s+", "", text)

    return text


target_contexts = {}
context_keys_by_query = defaultdict(list)

context_counts = []

for qid in selected_ids:

    row = original_by_id[qid]

    titles = row["context"]["title"]
    sentence_lists = row["context"]["sentences"]

    context_counts.append(len(titles))

    for title, sentences in zip(titles, sentence_lists):

        text = " ".join(sentences).strip()

        key = (
            canonical(title),
            canonical(text)
        )

        context_keys_by_query[qid].append(key)

        if key not in target_contexts:
            target_contexts[key] = {
                "title_en": title,
                "text_en": text
            }


print("Questions:", len(selected_ids))
print("Context occurrences:", sum(context_counts))
print("Unique contexts after dedup:", len(target_contexts))

print("\nContexts/query distribution:")
print(Counter(context_counts))
```

## Cell A3. Map các context sang tiếng Việt

```python
vn_corpus_stream = load_dataset(
    VN_REPO,
    "corpus",
    split="test",
    streaming=True
)

mapped_contexts = {}

target_keys = set(target_contexts.keys())

for i, row in enumerate(vn_corpus_stream, start=1):

    key = (
        canonical(row["og_title"]),
        canonical(row["og_text"])
    )

    if key in target_keys and key not in mapped_contexts:

        mapped_contexts[key] = {
            "id": str(row["id"]),
            "title": row["title"],
            "text": row["text"],
            "og_title": row["og_title"],
            "og_text": row["og_text"]
        }

    if len(mapped_contexts) == len(target_keys):
        print("Found all contexts!")
        break

    if i % 100000 == 0:
        print(
            f"Scanned {i:,} | "
            f"Found {len(mapped_contexts):,}/{len(target_keys):,}"
        )


print("\n" + "=" * 70)
print("CONTEXT MAPPING")
print("=" * 70)

print("Expected:", len(target_keys))
print("Found   :", len(mapped_contexts))
print("Missing :", len(target_keys) - len(mapped_contexts))

print(
    "Rate    :",
    f"{len(mapped_contexts)/len(target_keys)*100:.2f}%"
)
```

## Cell A4. Tạo shared Vietnamese corpus

```python
final_corpus = {}

for key, doc in mapped_contexts.items():
    final_corpus[doc["id"]] = doc

print("Final unique Vietnamese corpus docs:", len(final_corpus))
```

## Cell A5. Map từng query sang 10 context IDs

```python
context_doc_ids_by_query = defaultdict(list)

missing_context_links = []

for qid in selected_ids:

    for key in context_keys_by_query[qid]:

        doc = mapped_contexts.get(key)

        if doc is None:
            missing_context_links.append((qid, key))
        else:
            context_doc_ids_by_query[qid].append(
                doc["id"]
            )


print("Missing query-context mappings:",
      len(missing_context_links))

print("\nExample:")
qid = selected_ids[0]

print("Query:", vn_by_id[qid]["text"])
print("Number of contexts:",
      len(context_doc_ids_by_query[qid]))
print("IDs:", context_doc_ids_by_query[qid])
```

## Cell A6.

```python
selected_qrels = []

qrels_by_query = defaultdict(list)

for row in qrels:

    qid = str(row["query-id"])

    if qid in selected_id_set:

        cid = str(row["corpus-id"])

        selected_qrels.append({
            "query_id": qid,
            "corpus_id": cid,
            "score": int(row["score"])
        })

        qrels_by_query[qid].append(cid)


print("Selected qrels:", len(selected_qrels))

print(
    "Relevant docs/query:",
    Counter(
        len(x)
        for x in qrels_by_query.values()
    )
)
```

## Cell A7. Kiểm tra 2 gold docs có nằm trong final corpus không


```python
missing_gold = []

for row in selected_qrels:

    if row["corpus_id"] not in final_corpus:
        missing_gold.append(row)


print("Missing gold documents:", len(missing_gold))
```

# B. Bổ sung metadata từ HotpotQA gốc

## Cell B1. Tạo records cho 1.000 queries

```python
query_records = []

for qid in selected_ids:

    en = original_by_id[qid]
    vi = vn_by_id[qid]

    supporting_facts = []

    for title, sent_id in zip(
        en["supporting_facts"]["title"],
        en["supporting_facts"]["sent_id"]
    ):
        supporting_facts.append({
            "title": title,
            "sent_id": int(sent_id)
        })

    query_records.append({
        "id": qid,

        "question_vi": vi["text"],
        "question_en": en["question"],

        "answer_en": en["answer"],

        "type": en["type"],
        "level": en["level"],

        "gold_doc_ids": qrels_by_query[qid],

        "context_doc_ids":
            context_doc_ids_by_query[qid],

        "supporting_facts":
            supporting_facts
    })


print("Query records:", len(query_records))
print(query_records[0])
```

# C. Flow tạo answer_vi

## Cell C1 — Tạo text của 2 gold docs cho mỗi query

```python
gold_text_vi_by_query = {}

for qid in selected_ids:

    texts = []

    for cid in qrels_by_query[qid]:

        doc = final_corpus[cid]

        texts.append(
            f"{doc['title']}\n{doc['text']}"
        )

    gold_text_vi_by_query[qid] = "\n".join(texts)
```

## Cell C2 — Rule xử lý answer đơn giản trước

```python
def normalize_match(text):
    if text is None:
        return ""

    text = unicodedata.normalize(
        "NFKC",
        str(text)
    )

    text = text.casefold()
    text = re.sub(r"\s+", " ", text)

    return text.strip()


answer_results = {}
need_translation = []


for record in query_records:

    qid = record["id"]
    answer_en = record["answer_en"].strip()

    a = normalize_match(answer_en)

    gold_vi = normalize_match(
        gold_text_vi_by_query[qid]
    )

    # YES / NO
    if a == "yes":

        answer_results[qid] = {
            "answer_vi": "có",
            "method": "rule_yes_no",
            "confidence": "high",
            "needs_review": False
        }

    elif a == "no":

        answer_results[qid] = {
            "answer_vi": "không",
            "method": "rule_yes_no",
            "confidence": "high",
            "needs_review": False
        }

    # Nếu answer tiếng Anh vẫn xuất hiện nguyên văn
    # trong Vietnamese passage -> thường proper noun/number
    elif len(a) >= 3 and a in gold_vi:

        answer_results[qid] = {
            "answer_vi": answer_en,
            "method": "exact_span_existing",
            "confidence": "high",
            "needs_review": False
        }

    else:
        need_translation.append(record)


print("Solved automatically:",
      len(answer_results))

print("Need translation:",
      len(need_translation))
```

## C3 — Dịch các answer còn lại bằng NLLB

```python
import torch
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

MODEL_NAME = "facebook/nllb-200-distilled-600M"

device = (
    "cuda"
    if torch.cuda.is_available()
    else "cpu"
)

print("Device:", device)

tokenizer = AutoTokenizer.from_pretrained(
    MODEL_NAME
)

model = AutoModelForSeq2SeqLM.from_pretrained(
    MODEL_NAME
).to(device)

tokenizer.src_lang = "eng_Latn"

VI_TOKEN = tokenizer.convert_tokens_to_ids(
    "vie_Latn"
)
```

## Cell C4. Batch translation

```python
def translate_batch(texts, batch_size=32):

    outputs = []

    for start in range(
        0,
        len(texts),
        batch_size
    ):

        batch = texts[
            start:start + batch_size
        ]

        inputs = tokenizer(
            batch,
            return_tensors="pt",
            padding=True,
            truncation=True,
            max_length=64
        ).to(device)

        with torch.no_grad():

            generated = model.generate(
                **inputs,
                forced_bos_token_id=VI_TOKEN,
                max_new_tokens=32
            )

        translated = tokenizer.batch_decode(
            generated,
            skip_special_tokens=True
        )

        outputs.extend(translated)

    return outputs
```

```python
answers_to_translate = [
    r["answer_en"]
    for r in need_translation
]

translated_answers = translate_batch(
    answers_to_translate
)

print(
    "Translated:",
    len(translated_answers)
)
```

## Cell C5. Kiểm tra candidate có nằm trong gold Vietnamese evidence không

```python
for record, candidate in zip(
    need_translation,
    translated_answers
):

    qid = record["id"]

    candidate = candidate.strip()

    candidate_norm = normalize_match(candidate)

    gold_norm = normalize_match(
        gold_text_vi_by_query[qid]
    )

    if (
        len(candidate_norm) >= 2
        and candidate_norm in gold_norm
    ):

        confidence = "high"
        review = False
        method = "translated_exact_span"

    else:

        confidence = "medium"
        review = True
        method = "translated_non_extractive"

    answer_results[qid] = {
        "answer_vi": candidate,
        "method": method,
        "confidence": confidence,
        "needs_review": review
    }
```

## Cell C6 — Gắn answer_vi vào records

```python
final_queries = []

for record in query_records:

    result = answer_results[
        record["id"]
    ]

    new_record = record.copy()

    new_record.update({
        "answer_vi":
            result["answer_vi"],

        "answer_vi_method":
            result["method"],

        "answer_vi_confidence":
            result["confidence"],

        "answer_vi_needs_review":
            result["needs_review"]
    })

    final_queries.append(new_record)


print("Final queries:", len(final_queries))
```

## Cell C7.1 Load model

```python
import re
import unicodedata
import torch

from sentence_transformers import SentenceTransformer

CHECK_MODEL = "sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2"

check_model = SentenceTransformer(
    CHECK_MODEL,
    device="cuda" if torch.cuda.is_available() else "cpu"
)

print("Loaded:", CHECK_MODEL)
```

## Cell C7.2 Flexible normalization

```python
def flexible_norm(text):
    if text is None:
        return ""

    text = unicodedata.normalize("NFKC", str(text))
    text = text.casefold()

    # 27.000 / 27,000 -> 27000
    text = re.sub(r"(?<=\d)[.,](?=\d)", "", text)

    # punctuation -> space
    text = re.sub(
        r"[^\wÀ-ỹ]+",
        " ",
        text,
        flags=re.UNICODE
    )

    text = re.sub(r"\s+", " ", text)

    return text.strip()


review_records = [
    x for x in final_queries
    if x["answer_vi_needs_review"]
]

print("Review records:", len(review_records))
```

## Cell C7.3. Tính semantic similarity

```python
q_vi_list = [
    x["question_vi"]
    for x in review_records
]

q_en_list = [
    x["question_en"]
    for x in review_records
]

a_vi_list = [
    x["answer_vi"]
    for x in review_records
]

a_en_list = [
    x["answer_en"]
    for x in review_records
]


# Encode theo batch
q_vi_emb = check_model.encode(
    q_vi_list,
    batch_size=64,
    convert_to_tensor=True,
    normalize_embeddings=True
)

q_en_emb = check_model.encode(
    q_en_list,
    batch_size=64,
    convert_to_tensor=True,
    normalize_embeddings=True
)

a_vi_emb = check_model.encode(
    a_vi_list,
    batch_size=64,
    convert_to_tensor=True,
    normalize_embeddings=True
)

a_en_emb = check_model.encode(
    a_en_list,
    batch_size=64,
    convert_to_tensor=True,
    normalize_embeddings=True
)


question_sim = (
    q_vi_emb * q_en_emb
).sum(dim=1).cpu().tolist()

answer_sim = (
    a_vi_emb * a_en_emb
).sum(dim=1).cpu().tolist()


print("Encoded", len(review_records), "review cases")
```

## Cell C7.4. Kiểm tra answer có được gold context hỗ trợ không

```python
def split_sentences(text):
    parts = re.split(
        r"(?<=[.!?])\s+|\n+",
        str(text)
    )

    return [
        x.strip()
        for x in parts
        if len(x.strip()) >= 10
    ]


sentence_bank = []
sentence_ranges = {}

for record in review_records:

    qid = record["id"]

    sentences = split_sentences(
        gold_text_vi_by_query[qid]
    )

    start = len(sentence_bank)

    sentence_bank.extend(sentences)

    end = len(sentence_bank)

    sentence_ranges[qid] = (start, end)


print("Gold context sentences:", len(sentence_bank))


sentence_emb = check_model.encode(
    sentence_bank,
    batch_size=128,
    convert_to_tensor=True,
    normalize_embeddings=True,
    show_progress_bar=True
)
```

```python
context_sim = []
best_context_sentence = []

for i, record in enumerate(review_records):

    qid = record["id"]

    start, end = sentence_ranges[qid]

    if end <= start:

        context_sim.append(0.0)
        best_context_sentence.append("")
        continue

    candidate_emb = a_vi_emb[i]

    sims = (
        sentence_emb[start:end]
        @ candidate_emb
    )

    best_idx = int(
        torch.argmax(sims).item()
    )

    context_sim.append(
        float(sims[best_idx].item())
    )

    best_context_sentence.append(
        sentence_bank[start + best_idx]
    )
```

## Cell C7.5 — Chia 216 câu thành A/B/C/D

```python
triage_results = []

for i, record in enumerate(review_records):

    qid = record["id"]

    ans_vi = record["answer_vi"]

    gold_context = (
        gold_text_vi_by_query[qid]
    )

    flexible_span = (
        len(flexible_norm(ans_vi)) >= 2
        and flexible_norm(ans_vi)
        in flexible_norm(gold_context)
    )

    qs = question_sim[i]
    ans = answer_sim[i]
    ctx = context_sim[i]

    # -----------------------------------------
    # A: gần như ổn
    # -----------------------------------------
    if flexible_span:

        category = "A_flexible_span"

    # -----------------------------------------
    # C: nghi question VI dịch lệch question EN
    # -----------------------------------------
    elif qs < 0.60:

        category = "C_question_suspicious"

    # -----------------------------------------
    # B: nghi answer_vi dịch sai answer_en
    # -----------------------------------------
    elif ans < 0.50:

        category = "B_answer_translation_suspicious"

    # -----------------------------------------
    # D: answer translation có vẻ đúng
    # nhưng evidence VI không support tốt
    # -----------------------------------------
    elif ctx < 0.28:

        category = "D_context_support_suspicious"

    # -----------------------------------------
    # A: semantic đều tốt
    # -----------------------------------------
    elif (
        qs >= 0.72
        and ans >= 0.65
        and ctx >= 0.40
    ):

        category = "A_likely_ok"

    else:

        category = "E_ambiguous_review"


    triage_results.append({
        "id": qid,

        "question_vi":
            record["question_vi"],

        "question_en":
            record["question_en"],

        "answer_en":
            record["answer_en"],

        "answer_vi":
            record["answer_vi"],

        "question_similarity": qs,

        "answer_similarity": ans,

        "context_similarity": ctx,

        "best_context_sentence":
            best_context_sentence[i],

        "flexible_span":
            flexible_span,

        "category":
            category
    })
```

## Cell C7.7 — Tạo gold supporting context tiếng Anh

Dùng đúng các supporting_facts của HotpotQA gốc.

```python
def build_gold_en_context(qid):
    row = original_by_id[qid]

    title_to_sentences = {
        title: sentences
        for title, sentences in zip(
            row["context"]["title"],
            row["context"]["sentences"]
        )
    }

    support = []

    for title, sent_id in zip(
        row["supporting_facts"]["title"],
        row["supporting_facts"]["sent_id"]
    ):
        if (
            title in title_to_sentences
            and sent_id < len(title_to_sentences[title])
        ):
            sent = title_to_sentences[title][sent_id]
            support.append(f"{title}: {sent}")

    return " ".join(support)


gold_en_context_by_query = {
    x["id"]: build_gold_en_context(x["id"])
    for x in final_queries
}

print("Built contexts:", len(gold_en_context_by_query))
```

## Cell C7.8 Load model kiểm tra QA consistency


```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
import torch

QA_CHECK_MODEL = "google/flan-t5-base"

qa_tokenizer = AutoTokenizer.from_pretrained(QA_CHECK_MODEL)

qa_model = AutoModelForSeq2SeqLM.from_pretrained(
    QA_CHECK_MODEL
).to(
    "cuda" if torch.cuda.is_available() else "cpu"
)

qa_device = next(qa_model.parameters()).device

print("QA checker:", QA_CHECK_MODEL)
print("Device:", qa_device)
```

## Cell C7.9 — Kiểm tra 1.000 gold answers

```python
def make_consistency_prompt(record):

    context = gold_en_context_by_query[record["id"]]

    return f"""
Determine whether the proposed answer directly and correctly answers
the question using the context.

Important:
- The answer must answer exactly what the question asks.
- Do NOT say YES merely because the proposed answer appears in the context.
- Check question type carefully: who, what, why, when, where, which, how many, etc.
- Reply with only YES or NO.

Context:
{context}

Question:
{record["question_en"]}

Proposed answer:
{record["answer_en"]}

Correct?
""".strip()
```

```python
def run_qa_consistency(records, batch_size=16):

    results = []

    for start in range(0, len(records), batch_size):

        batch_records = records[start:start + batch_size]

        prompts = [
            make_consistency_prompt(x)
            for x in batch_records
        ]

        inputs = qa_tokenizer(
            prompts,
            return_tensors="pt",
            padding=True,
            truncation=True,
            max_length=512
        ).to(qa_device)

        with torch.no_grad():
            outputs = qa_model.generate(
                **inputs,
                max_new_tokens=4
            )

        texts = qa_tokenizer.batch_decode(
            outputs,
            skip_special_tokens=True
        )

        results.extend(texts)

    return results


qa_judgments_raw = run_qa_consistency(
    final_queries
)

print("Checked:", len(qa_judgments_raw))
```

## Cell C7.10 — Parse kết quả

```python
qa_consistency = {}

for record, raw in zip(
    final_queries,
    qa_judgments_raw
):

    text = raw.strip().lower()

    if text.startswith("yes"):
        status = "pass"

    elif text.startswith("no"):
        status = "suspicious"

    else:
        status = "uncertain"

    qa_consistency[record["id"]] = {
        "status": status,
        "raw": raw
    }


from collections import Counter

print(
    Counter(
        x["status"]
        for x in qa_consistency.values()
    )
)
```

## Cell C7.12 — Sinh answer tham khảo cho những câu bị nghi

```python
source_suspicious = [
    x for x in final_queries
    if qa_consistency[x["id"]]["status"] != "pass"
]

print("Source QA cases to inspect:",
      len(source_suspicious))
```

```python
def make_answer_prompt(record):

    context = gold_en_context_by_query[record["id"]]

    return f"""
Answer the question using only the context.
Return only a short answer.

Context:
{context}

Question:
{record["question_en"]}

Answer:
""".strip()


def generate_reference_answers(records, batch_size=16):

    answers = []

    for start in range(0, len(records), batch_size):

        batch = records[start:start + batch_size]

        prompts = [
            make_answer_prompt(x)
            for x in batch
        ]

        inputs = qa_tokenizer(
            prompts,
            return_tensors="pt",
            padding=True,
            truncation=True,
            max_length=512
        ).to(qa_device)

        with torch.no_grad():

            outputs = qa_model.generate(
                **inputs,
                max_new_tokens=32
            )

        answers.extend(
            qa_tokenizer.batch_decode(
                outputs,
                skip_special_tokens=True
            )
        )

    return answers


reference_answers = generate_reference_answers(
    source_suspicious
)

reference_answer_by_id = {
    row["id"]: ans
    for row, ans in zip(
        source_suspicious,
        reference_answers
    )
}
```
